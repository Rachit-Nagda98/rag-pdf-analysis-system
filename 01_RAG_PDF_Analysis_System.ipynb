### STEP 0: CREATION OF LLM = ChatOpenAI

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-5-nano", temperature=0)
```

### STEP 1: Extracting Text from PDF using PyPDFLoader

```python
from langchain_community.document_loaders import PyPDFLoader
pdf_path = "../Docs/Animal_Planet_Detailed_Book.pdf"
loader=PyPDFLoader(pdf_path)
docs=loader.load()
docs
```

### STEP 2: Creating Chunks of our document

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=20
)
chunks = splitter.split_documents(docs)
chunks
```

### STEP 3: creation of embedding model

```python
from langchain_openai import OpenAIEmbeddings

embedding_model = OpenAIEmbeddings(model = "text-embedding-3-small")
```

### STEP 4: STORE EMBEDDINGS IN TO VECTOR DATABASE

```python
from langchain_community.vectorstores import Chroma

vectorstore = Chroma.from_documents(
    documents = chunks,
    embedding = embedding_model,
    persist_directory="../Vector/"
)
```

### STEP 5: SEMENTIC SEARCH

```python
response = vectorstore.similarity_search("what is dolphin", k=3)
response
```
